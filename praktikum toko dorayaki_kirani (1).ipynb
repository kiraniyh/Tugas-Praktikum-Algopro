praktikum: Toko Dorayaki Doraemon

program ini dibuat untuk simulasi sederhana toko dorayaki Doraemon. program menerima input nama makanan, nama karakter, dan kata sifat untuk nama toko. lalu, program menghitung modal dari Doraemon, Nobita, dan Shizuka, menghitung pendapatan penjualan dorayaki, dan menghitung keuntungan yang mereka dapat dari bulan pertama.

```python
print("nama toko?") # menampilkan pertanyaan nama toko
# input nama makanan
nama_makan = input("masukan nama makan:")
nama_karakter = input("masukan nama karakter:")
kata_sifat = input("masukan kata sifat:")
# menggabungkan tiga variabel
print("nama resmi toko:", nama_makan, nama_karakter, kata_sifat)
# modal dari masing-masing karakter
modaldoraemon = 555000
modalnobita = 333000
modalshizuka = 222000
# menghitung total modal yang didapat
total_modal = modaldoraemon + modalnobita + modalshizuka
# menampilkan total modal
print("total modal untuk membuka toko dorayaki =", total_modal)
# variabel pendapatan
jumlah_dorayaki = 350
harga_dorayaki = 5000
# menghitung pendapatan
pendapatan = jumlah_dorayaki * harga_dorayaki
# menampilkan hasil pendapatan
print("pendapatan bulan pertama=", pendapatan)
# menghitung keuntungan
keuntungan = pendapatan - total_modal
# menampilkan keuntungan
print("keuntungan total adalah=", keuntungan)
```
