## program Smart Vehicle Insurance Analyzer

```python
# Doraemon menentukan kategori risiko dan premi asuransi Nobita

usia_pengemudi = int(input("Masukkan usia pengemudi: "))
jenis_kendaraan = input("Masukkan jenis kendaraan (mobil/motor): ").lower()
nilai_kendaraan = float(input("Masukkan nilai kendaraan (Rp): "))
area_penggunaan = input("Masukkan area penggunaan (perkotaan/pedesaan): ").lower()
sistem_pengaman = input("Apakah memiliki sistem pengaman? (ya/tidak): ").lower()

# kategori risiko berdasarkan usia
if usia_pengemudi < 20 or usia_pengemudi > 60:
    risiko = 3
else:
    risiko = 1

# risiko berdasarkan jenis kendaraan
if jenis_kendaraan == "motor":
    risiko += 1
elif jenis_kendaraan == "mobil":
    risiko += 0

# risiko berdasarkan area penggunaan
if area_penggunaan == "perkotaan":
    risiko += 1

#  risiko berdasarkan sistem pengaman
if sistem_pengaman == "tidak":
    risiko += 1

# kategori risiko
if risiko >= 4:
    kategori = "Tinggi"
    persentase_premi = 0.1
elif risiko >= 2:
    kategori = "Sedang"
    persentase_premi = 0.07
else:
    kategori = "Rendah"
    persentase_premi = 0.05

# Menghitung premi asuransi
premi = nilai_kendaraan * persentase_premi

# pesan Doraemon conditional expression
pesan = (
    "Doraemon menyarankan asuransi tambahan untuk perlindungan penuh!"
    if kategori == "Tinggi"
    else "Risiko kendaraan masih aman menurut Doraemon."
)

# hasil
print("\n======= HASIL ASURANSI =======")
print("Kategori risiko:", kategori)
print(f"Premi asuransi: Rp{premi:,.0f}".replace(",", "."))
print("Pesan Doraemon:", pesan)
```
