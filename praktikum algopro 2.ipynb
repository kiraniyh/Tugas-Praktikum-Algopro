```python
target_minum = 2000 
ukuran_botol = 350

hasil = target_minum / ukuran_botol
botol_penuh = target_minum // ukuran_botol
sisa_air = target_minum % ukuran_botol

print("hasil true division:", hasil)
print("botol penuh:", botol_penuh)
print("sisa air:", sisa_air, "ml")
```
