```python
belanjaan = ["apel", "susu", "roti"]
harga_jumlah = (5000, 3, 12000, 2, 8000, 1)

print("list belanjaan nobita")
print(belanjaan)

print("\nkarakter pertama, barang pertama:", belanjaan[0][0])
print("karakter terakhir, barang terakhir:", belanjaan[-1][-1])

total_apel = harga_jumlah[0] * harga_jumlah[1]
total_susu = harga_jumlah[2] * harga_jumlah[3]
total_roti = harga_jumlah[4] * harga_jumlah[5]


print("\ntotal harga tiap barang:")
print("Apel:", total_apel)
print("Susu:", total_susu)
print("Roti:", total_roti)

total_belanjaan = total_apel + total_susu + total_roti

print("total seluruh belanjaan:", total_belanjaan)

pembeli = "Nobita"
toko = "toko sembako doraemon"


print("\n" + "=" * 40)
print(f"{toko:^40}")
print("=" * 40)
print(f"Pembeli: {pembeli}")
print("-" * 40)
print(f"{'Barang':<10}{'Harga':>10}{'Jumlah':>8}{'Total':>10}")
print("-" * 40)

print(f"{belanjaan[0]:<10}{harga_jumlah[0]:>10}{harga_jumlah[1]:>8}{total_apel:>10}")
print(f"{belanjaan[1]:<10}{harga_jumlah[2]:>10}{harga_jumlah[3]:>8}{total_susu:>10}")
print(f"{belanjaan[2]:<10}{harga_jumlah[4]:>10}{harga_jumlah[5]:>8}{total_roti:>10}")

print("-" * 40)
print(f"{'Total Belanja':<30}{total_belanjaan:>10}")
print("=" * 40)
print(f"{'Terima kasih sudah belanja!':^40}")
print("=" * 40)

belanjaan.append("Pisang")

print("\nList Belanjaan Setelah Ditambah Pisang:")
print(belanjaan)
```
