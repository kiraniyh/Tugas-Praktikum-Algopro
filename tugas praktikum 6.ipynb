```python

# a. Membuat Variabel Global
skor_nobita = 0


# b. Membuat Fungsi untuk Memperbarui Skor
def update_skor(nilai):
    global skor_nobita
    skor_nobita += nilai


# c. Membuat Fungsi Aktivitas Nobita

# 1. Fungsi Belajar
def belajar(jam_belajar):
    if jam_belajar < 1:
        skor = 50
        print("Nobita hanya membuka buku sebentar")
    elif jam_belajar < 3:
        skor = 25
        print("Nobita belajar secukupnya")
    else:
        skor = 10
        print("Nobita belajar dengan rajin hari ini")

    return skor


# 2. Fungsi Tidur
def tidur(jam_tidur):
    if jam_tidur < 1:
        skor = -10
        print("Nobita kurang tidur")
    elif jam_tidur <= 3:
        skor = 20
        print("Waktu tidur Nobita seimbang")
    else:
        skor = -20
        print("Nobita tidur terlalu lama")

    return skor


# 3. Fungsi Main Bersama Doraemon
def main_bersama(jumlah_gadget, waktu_main):
    if jumlah_gadget <= 2 and waktu_main <= 1:
        skor = 15
        print("Nobita bermain secukupnya")
    elif jumlah_gadget <= 3 and waktu_main <= 2:
        skor = 5
        print("Nobita masih bisa mengatur waktu dengan baik")
    else:
        skor = -30
        print("Nobita kebanyakan bermain")

    return skor


# 4. Fungsi Menyontek
def menyontek():
    print("Nobita ketahuan menyontek")
    return -50


# d. Membuat Fungsi Pengendali Aktivitas
def pengendali_aktivitas(pilihan):

    if pilihan == 1:
        jam = float(input("Berapa jam Nobita belajar hari ini? "))
        nilai = belajar(jam)
        update_skor(nilai)

    elif pilihan == 2:
        jam = float(input("Berapa jam Nobita tidur siang hari ini? "))
        nilai = tidur(jam)
        update_skor(nilai)

    elif pilihan == 3:
        gadget = int(input("Berapa gadget yang digunakan Nobita? "))
        waktu = float(input("Berapa jam Nobita bermain gadget? "))
        nilai = main_bersama(gadget, waktu)
        update_skor(nilai)

    elif pilihan == 4:
        nilai = menyontek()
        update_skor(nilai)

    else:
        print("Pilihan tidak valid, aktivitas hanya ada 4 pilihan")


# e. Membuat Fungsi Evaluasi Akhir
def evaluasi_akhir():
    print("\n=== HASIL EVALUASI NOBITA ===")
    print("Total Skor Nobita:", skor_nobita)

    if skor_nobita >= 80:
        print("Siswa Teladan")
    elif skor_nobita >= 50:
        print("Cukup Rajin")
    else:
        print("Pemalas")


# f. Membuat Menu Utama Program
print("=== PROGRAM PENINGKAT DISIPLIN NOBITA ===")
print("1. Belajar")
print("2. Tidur")
print("3. Main bersama Doraemon")
print("4. Menyontek saat ujian")

pilihan = int(input("\nPilih aktivitas (1-4): "))

if 1 <= pilihan <= 4:
    pengendali_aktivitas(pilihan)
    evaluasi_akhir()
else:
    pengendali_aktivitas(pilihan)
```
